# Google Maps 车行联系方式抓取（Places API 版）

- 使用 **Places Nearby Search** + **Place Details** 获取车行名称、地址、电话、官网等。
- 先配置 `API_KEY` 和目标 `CITY_NAME`/`location`（经纬度）。
- 结果输出为 Excel/CSV。


```python
import time
import requests
import pandas as pd
from urllib.parse import urlencode

# ====== 基础配置 ======
API_KEY = "YOUR_API_KEY"  # TODO: 替换成你的 Google Places API Key
CITY_NAME = ""  # 可选：城市名称（仅用于输出标识）
LOCATION = ""  # 必填：纬度,经度，例如 "31.2304,121.4737"
RADIUS = 5000  # 搜索半径（米），可按需调整
KEYWORD = "car dealer"  # 关键词，可用: car dealer / auto dealer / 汽车经销商
TYPE = "car_dealer"

if not API_KEY or not LOCATION:
    raise ValueError("请先填写 API_KEY 和 LOCATION（纬度,经度）")

# ====== Nearby Search 获取 Place IDs ======
nearby_url = "https://maps.googleapis.com/maps/api/place/nearbysearch/json"
params = {
    "location": LOCATION,
    "radius": RADIUS,
    "keyword": KEYWORD,
    "type": TYPE,
    "key": API_KEY,
}

place_ids = []
while True:
    resp = requests.get(nearby_url, params=params, timeout=20)
    data = resp.json()
    place_ids += [r.get("place_id") for r in data.get("results", []) if r.get("place_id")]

    token = data.get("next_page_token")
    if not token:
        break
    time.sleep(2)  # next_page_token 需要等待
    params["pagetoken"] = token

# 去重
place_ids = list(dict.fromkeys(place_ids))
print(f"✅ 共抓到 {len(place_ids)} 个 Place IDs")

# ====== Place Details 获取联系方式 ======
details_url = "https://maps.googleapis.com/maps/api/place/details/json"
fields = [
    "name",
    "formatted_address",
    "formatted_phone_number",
    "international_phone_number",
    "website",
    "rating",
    "user_ratings_total",
]

rows = []
for i, pid in enumerate(place_ids, 1):
    dparams = {
        "place_id": pid,
        "fields": ",".join(fields),
        "key": API_KEY,
    }
    dresp = requests.get(details_url, params=dparams, timeout=20)
    ddata = dresp.json().get("result", {})
    rows.append({
        "城市": CITY_NAME,
        "名称": ddata.get("name"),
        "地址": ddata.get("formatted_address"),
        "电话": ddata.get("formatted_phone_number"),
        "国际电话": ddata.get("international_phone_number"),
        "官网": ddata.get("website"),
        "评分": ddata.get("rating"),
        "评分人数": ddata.get("user_ratings_total"),
        "place_id": pid,
    })
    if i % 20 == 0:
        print(f"…已处理 {i}/{len(place_ids)}")
    time.sleep(0.1)

df = pd.DataFrame(rows)
df.to_excel("car_dealers_contacts.xlsx", index=False)
df.to_csv("car_dealers_contacts.csv", index=False)
print("✅ 已导出 car_dealers_contacts.xlsx / .csv")
```
